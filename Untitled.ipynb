```python
# ============================================================
# CELL 8 — GAUSSIAN KERNEL
# ============================================================

def gaussian_kernel(
    kernel_size=5,
    sigma=1.0,
    device="cpu"
):

    radius = kernel_size // 2

    x = torch.arange(
        -radius,
        radius + 1,
        dtype=torch.float32,
        device=device
    )

    kernel_1d = torch.exp(
        -(x ** 2) /
        (2 * sigma ** 2)
    )

    kernel_2d = torch.outer(
        kernel_1d,
        kernel_1d
    )

    kernel_2d /= kernel_2d.sum()

    return kernel_2d.view(
        1,
        1,
        kernel_size,
        kernel_size
    )


kernel = gaussian_kernel()

print(kernel)

print(
    "Kernel sum:",
    kernel.sum().item()
)
```

```python
# ============================================================
# CELL 9 — HR → LR DEM DEGRADATION
# ============================================================

def degrade_dem(
    gt,
    scale_factor=5
):

    """
    HR DEM → blurred/downsampled LR DEM

    Input:
        [B, 1, H, W]

    Output:
        [B, 1, H/scale, W/scale]
    """

    kernel = gaussian_kernel(
        kernel_size=5,
        sigma=1.0,
        device=gt.device
    )

    pad = kernel.shape[-1] // 2

    # Reflect padding prevents artificial zero borders
    padded = F.pad(
        gt,
        (pad, pad, pad, pad),
        mode="reflect"
    )

    blurred = F.conv2d(
        padded,
        kernel
    )

    lr = F.avg_pool2d(
        blurred,
        kernel_size=scale_factor,
        stride=scale_factor
    )

    return lr


print("✓ Degradation function ready")
```

```python
# ============================================================
# CELL 11 — LUNAR DEM DATASET
# ============================================================

class LunarDEMDataset(Dataset):

    def __init__(
        self,
        geotiff_path,
        hr_size=100,
        scale_factor=5,
        num_patches=5000,
        split="train"
    ):

        self.hr_size = hr_size

        self.scale_factor = scale_factor

        self.num_patches = num_patches

        self.split = split

        # ----------------------------------------------------
        # Load DEM
        # ----------------------------------------------------

        with rasterio.open(
            geotiff_path
        ) as src:

            self.dem = src.read(
                1
            ).astype(
                np.float32
            )

        # ----------------------------------------------------
        # Clean invalid values
        # ----------------------------------------------------

        median_value = np.nanmedian(
            self.dem
        )

        self.dem = np.nan_to_num(
            self.dem,
            nan=median_value,
            posinf=median_value,
            neginf=median_value
        )

        self.height, self.width = (
            self.dem.shape
        )

        # ----------------------------------------------------
        # Spatial split
        # ----------------------------------------------------

        train_end = int(
            self.height * 0.70
        )

        val_end = int(
            self.height * 0.85
        )

        if split == "train":

            self.y_min = 0
            self.y_max = train_end

        elif split == "val":

            self.y_min = train_end
            self.y_max = val_end

        elif split == "test":

            self.y_min = val_end
            self.y_max = self.height

        else:

            raise ValueError(
                "split must be train, val, or test"
            )

        # ----------------------------------------------------
        # Valid patch coordinate range
        # ----------------------------------------------------

        max_y = self.y_max - hr_size
        max_x = self.width - hr_size

        if max_y < self.y_min:

            raise ValueError(
                "Region too small for requested patch size."
            )

        # ----------------------------------------------------
        # Coordinates
        # ----------------------------------------------------

        if split == "train":

            rng = np.random.default_rng()

        else:

            rng = np.random.default_rng(
                SEED
            )

        self.coordinates = []

        for _ in range(num_patches):

            y = rng.integers(
                self.y_min,
                max_y + 1
            )

            x = rng.integers(
                0,
                max_x + 1
            )

            self.coordinates.append(
                (y, x)
            )

    def __len__(self):

        return len(
            self.coordinates
        )

    def __getitem__(self, idx):

        y, x = self.coordinates[idx]

        # ----------------------------------------------------
        # Extract HR patch
        # ----------------------------------------------------

        gt = self.dem[
            y:y + self.hr_size,
            x:x + self.hr_size
        ]

        gt = torch.from_numpy(
            gt.copy()
        ).float()

        gt = gt.unsqueeze(0)

        # ----------------------------------------------------
        # Augmentation ONLY during training
        # ----------------------------------------------------

        if self.split == "train":

            if torch.rand(1).item() < 0.5:

                gt = torch.flip(
                    gt,
                    dims=[-1]
                )

            if torch.rand(1).item() < 0.5:

                gt = torch.flip(
                    gt,
                    dims=[-2]
                )

            k = torch.randint(
                0,
                4,
                (1,)
            ).item()

            gt = torch.rot90(
                gt,
                k=k,
                dims=[-2, -1]
            )

        # ----------------------------------------------------
        # Generate LR observation
        # ----------------------------------------------------

        lr = degrade_dem(
            gt.unsqueeze(0),
            self.scale_factor
        ).squeeze(0)

        # ----------------------------------------------------
        # Per-patch mean normalization
        # ----------------------------------------------------

        patch_mean = lr.mean()

        lr_norm = lr - patch_mean

        gt_norm = gt - patch_mean

        return (
            lr_norm,
            gt_norm,
            patch_mean
        )
```

```python
# ============================================================
# CELL 17 — RESIDUAL BLOCK
# ============================================================

class ResidualBlock(nn.Module):

    def __init__(
        self,
        channels,
        res_scale=0.1
    ):

        super().__init__()

        self.res_scale = res_scale

        self.conv1 = nn.Conv2d(
            channels,
            channels,
            kernel_size=3,
            padding=1
        )

        self.relu = nn.ReLU(
            inplace=True
        )

        self.conv2 = nn.Conv2d(
            channels,
            channels,
            kernel_size=3,
            padding=1
        )

    def forward(self, x):

        residual = self.conv1(x)

        residual = self.relu(
            residual
        )

        residual = self.conv2(
            residual
        )

        return (
            x +
            self.res_scale * residual
        )
```

```python
# ============================================================
# CELL 18 — FAST EDSR MODEL
# ============================================================

class FastEDSR(nn.Module):

    def __init__(
        self,
        scale_factor=5,
        num_features=48,
        num_blocks=4
    ):

        super().__init__()

        self.scale_factor = (
            scale_factor
        )

        # ----------------------------------------------------
        # Input: 1-channel DEM
        # ----------------------------------------------------

        self.head = nn.Conv2d(
            1,
            num_features,
            kernel_size=3,
            padding=1
        )

        # ----------------------------------------------------
        # Residual body
        # ----------------------------------------------------

        self.body = nn.Sequential(
            *[
                ResidualBlock(
                    num_features,
                    res_scale=0.1
                )
                for _ in range(num_blocks)
            ]
        )

        # ----------------------------------------------------
        # Body reconstruction
        # ----------------------------------------------------

        self.body_conv = nn.Conv2d(
            num_features,
            num_features,
            kernel_size=3,
            padding=1
        )

        # ----------------------------------------------------
        # IMPORTANT:
        # 48 features → 1 DEM channel
        # ----------------------------------------------------

        self.tail = nn.Conv2d(
            num_features,
            1,
            kernel_size=3,
            padding=1
        )

    def forward(self, x):

        # ----------------------------------------------------
        # Bilinear baseline
        # ----------------------------------------------------

        base = F.interpolate(
            x,
            scale_factor=self.scale_factor,
            mode="bilinear",
            align_corners=False
        )

        # ----------------------------------------------------
        # Feature extraction
        # ----------------------------------------------------

        feat = self.head(x)

        # ----------------------------------------------------
        # Residual processing
        # ----------------------------------------------------

        residual = self.body(
            feat
        )

        residual = self.body_conv(
            residual
        )

        # ----------------------------------------------------
        # Convert to one elevation channel
        # ----------------------------------------------------

        residual = self.tail(
            residual
        )

        # ----------------------------------------------------
        # Upsample ONLY one channel
        # ----------------------------------------------------

        residual = F.interpolate(
            residual,
            scale_factor=self.scale_factor,
            mode="bilinear",
            align_corners=False
        )

        # ----------------------------------------------------
        # Base + learned terrain detail
        # ----------------------------------------------------

        return (
            base +
            residual
        )
```

```python
# ============================================================
# CELL 22 — NORMALIZED SOBEL KERNELS
# ============================================================

sobel_x = torch.tensor(
    [
        [-1., 0., 1.],
        [-2., 0., 2.],
        [-1., 0., 1.]
    ],
    dtype=torch.float32
) / 8.0

sobel_y = torch.tensor(
    [
        [-1., -2., -1.],
        [ 0.,  0.,  0.],
        [ 1.,  2.,  1.]
    ],
    dtype=torch.float32
) / 8.0

print(sobel_x)
print(sobel_y)
```

```python
# ============================================================
# CELL 23 — PHYSICS-AWARE LOSS
# ============================================================

class LunarPhysicsLoss(nn.Module):

    def __init__(
        self,
        lambda_height=1.0,
        lambda_gradient=0.15,
        lambda_mean=0.05
    ):

        super().__init__()

        self.lambda_height = (
            lambda_height
        )

        self.lambda_gradient = (
            lambda_gradient
        )

        self.lambda_mean = (
            lambda_mean
        )

        self.register_buffer(
            "sobel_x",
            sobel_x.view(
                1, 1, 3, 3
            )
        )

        self.register_buffer(
            "sobel_y",
            sobel_y.view(
                1, 1, 3, 3
            )
        )

    def charbonnier(
        self,
        prediction,
        target,
        epsilon=1e-3
    ):

        diff = (
            prediction -
            target
        )

        return torch.mean(
            torch.sqrt(
                diff * diff +
                epsilon * epsilon
            )
        )

    def gradient_loss(
        self,
        prediction,
        target
    ):

        px = F.conv2d(
            prediction,
            self.sobel_x,
            padding=1
        )

        py = F.conv2d(
            prediction,
            self.sobel_y,
            padding=1
        )

        tx = F.conv2d(
            target,
            self.sobel_x,
            padding=1
        )

        ty = F.conv2d(
            target,
            self.sobel_y,
            padding=1
        )

        return (
            F.l1_loss(px, tx)
            +
            F.l1_loss(py, ty)
        )

    def forward(
        self,
        prediction,
        target
    ):

        height_loss = (
            self.charbonnier(
                prediction,
                target
            )
        )

        gradient_loss = (
            self.gradient_loss(
                prediction,
                target
            )
        )

        mean_loss = F.l1_loss(
            prediction.mean(
                dim=(-2, -1)
            ),
            target.mean(
                dim=(-2, -1)
            )
        )

        total_loss = (
            self.lambda_height *
            height_loss
            +
            self.lambda_gradient *
            gradient_loss
            +
            self.lambda_mean *
            mean_loss
        )

        return total_loss
```

```python
# ============================================================
# CELL 25 — TERRAIN ANALYTICS
# ============================================================

class TerrainAnalytics(nn.Module):

    def __init__(
        self,
        cell_size=1.0
    ):

        super().__init__()

        self.cell_size = (
            cell_size
        )

        self.register_buffer(
            "sobel_x",
            sobel_x.view(
                1, 1, 3, 3
            )
        )

        self.register_buffer(
            "sobel_y",
            sobel_y.view(
                1, 1, 3, 3
            )
        )

    def forward(self, dem):

        # ----------------------------------------------------
        # Elevation derivatives
        # ----------------------------------------------------

        dz_dx = F.conv2d(
            dem,
            self.sobel_x,
            padding=1
        ) / self.cell_size

        dz_dy = F.conv2d(
            dem,
            self.sobel_y,
            padding=1
        ) / self.cell_size

        # ----------------------------------------------------
        # Slope
        # ----------------------------------------------------

        gradient_magnitude = torch.sqrt(
            dz_dx ** 2 +
            dz_dy ** 2 +
            1e-8
        )

        slope = torch.rad2deg(
            torch.atan(
                gradient_magnitude
            )
        )

        # ----------------------------------------------------
        # Roughness
        # ----------------------------------------------------

        local_mean = F.avg_pool2d(
            dem,
            kernel_size=3,
            stride=1,
            padding=1
        )

        local_variance = F.avg_pool2d(
            (
                dem -
                local_mean
            ) ** 2,
            kernel_size=3,
            stride=1,
            padding=1
        )

        roughness = torch.sqrt(
            local_variance +
            1e-8
        )

        return (
            slope,
            roughness
        )


analytics = TerrainAnalytics(
    cell_size=1.0
).to(device)

print("✓ Terrain analytics ready")
```

```python
# ============================================================
# CELL 26 — HAZARD CLASSIFICATION
# ============================================================

def hazard_classes(
    slope
):

    hazard = torch.zeros_like(
        slope,
        dtype=torch.long
    )

    # Moderate
    hazard[
        slope >= SAFE_SLOPE
    ] = 1

    # Hazardous
    hazard[
        slope >= MODERATE_SLOPE
    ] = 2

    return hazard


print(
    f"Class 0: Safe < {SAFE_SLOPE}°"
)

print(
    f"Class 1: Moderate "
    f"{SAFE_SLOPE}–{MODERATE_SLOPE}°"
)

print(
    f"Class 2: Hazardous "
    f">= {MODERATE_SLOPE}°"
)
```

```python
# ============================================================
# CELL 28 — BILINEAR BASELINE
# ============================================================

def bilinear_baseline(
    lr,
    scale_factor=5
):

    return F.interpolate(
        lr,
        scale_factor=scale_factor,
        mode="bilinear",
        align_corners=False
    )


print("✓ Bilinear baseline ready")
```

```python
# ============================================================
# CELL 29 — HEIGHT METRICS
# ============================================================

def calculate_height_metrics(
    prediction,
    target
):

    rmse = torch.sqrt(
        torch.mean(
            (prediction - target) ** 2
        )
    )

    mae = torch.mean(
        torch.abs(
            prediction - target
        )
    )

    return (
        rmse.item(),
        mae.item()
    )
```

```python
# ============================================================
# CELL 30 — HAZARD METRICS
# ============================================================

def calculate_hazard_metrics(
    y_true,
    y_pred
):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="macro",
        zero_division=0
    )

    cm = confusion_matrix(
        y_true,
        y_pred
    )

    return {
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "confusion_matrix": cm
    }
```

```python
# ============================================================
# CELL 31 — VALIDATION
# ============================================================

def validate(
    model,
    loader,
    criterion,
    device
):

    model.eval()

    total_loss = 0.0

    with torch.no_grad():

        for lr, gt, mean in loader:

            lr = lr.to(
                device,
                non_blocking=True
            )

            gt = gt.to(
                device,
                non_blocking=True
            )

            with torch.autocast(
                device_type="cuda",
                enabled=(device.type == "cuda")
            ):

                prediction = model(
                    lr
                )

                loss = criterion(
                    prediction,
                    gt
                )

            total_loss += (
                loss.item()
            )

    return (
        total_loss /
        len(loader)
    )
```

```python
# ============================================================
# CELL 32 — OPTIMIZER
# ============================================================

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    weight_decay=WEIGHT_DECAY
)

print(
    "Optimizer:",
    optimizer.__class__.__name__
)
```

```python
# ============================================================
# CELL 33 — LEARNING RATE SCHEDULER
# ============================================================

scheduler = (
    torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer,
        T_max=EPOCHS,
        eta_min=1e-6
    )
)

print("✓ Cosine scheduler created")
```

```python
# ============================================================
# CELL 34 — MIXED PRECISION
# ============================================================

# New PyTorch API
scaler = torch.amp.GradScaler(
    "cuda",
    enabled=(device.type == "cuda")
)

print(
    "AMP enabled:",
    device.type == "cuda"
)
```

```python
# ============================================================
# CELL 35 — TRAIN LUNAR-SAFE V2
# ============================================================

train_history = []

val_history = []

lr_history = []

best_val_loss = float("inf")

BEST_MODEL_PATH = os.path.join(
    MODEL_DIR,
    "lunar_safe_v2_best.pth"
)

print("=" * 70)
print("STARTING LUNAR-SAFE V2 TRAINING")
print("=" * 70)

for epoch in range(
    1,
    EPOCHS + 1
):

    # --------------------------------------------------------
    # TRAIN
    # --------------------------------------------------------

    model.train()

    running_loss = 0.0

    epoch_start = time.time()

    for lr, gt, mean in train_loader:

        lr = lr.to(
            device,
            non_blocking=True
        )

        gt = gt.to(
            device,
            non_blocking=True
        )

        optimizer.zero_grad(
            set_to_none=True
        )

        # ----------------------------------------------------
        # Mixed precision forward
        # ----------------------------------------------------

        with torch.autocast(
            device_type="cuda",
            enabled=(device.type == "cuda")
        ):

            prediction = model(
                lr
            )

            loss = criterion(
                prediction,
                gt
            )

        # ----------------------------------------------------
        # Backpropagation
        # ----------------------------------------------------

        scaler.scale(
            loss
        ).backward()

        scaler.unscale_(
            optimizer
        )

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=1.0
        )

        scaler.step(
            optimizer
        )

        scaler.update()

        running_loss += (
            loss.item()
        )

    train_loss = (
        running_loss /
        len(train_loader)
    )

    # --------------------------------------------------------
    # VALIDATION
    # --------------------------------------------------------

    val_loss = validate(
        model,
        val_loader,
        criterion,
        device
    )

    # --------------------------------------------------------
    # Scheduler
    # --------------------------------------------------------

    scheduler.step()

    current_lr = (
        optimizer.param_groups[0]["lr"]
    )

    train_history.append(
        train_loss
    )

    val_history.append(
        val_loss
    )

    lr_history.append(
        current_lr
    )

    epoch_time = (
        time.time() -
        epoch_start
    )

    # --------------------------------------------------------
    # Print
    # --------------------------------------------------------

    print(
        f"Epoch "
        f"{epoch:03d}/{EPOCHS} | "
        f"Train: {train_loss:.6f} | "
        f"Val: {val_loss:.6f} | "
        f"LR: {current_lr:.2e} | "
        f"Time: {epoch_time:.1f}s"
    )

    # --------------------------------------------------------
    # Save best model
    # --------------------------------------------------------

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        torch.save(
            {
                "epoch": epoch,
                "model_state_dict": (
                    model.state_dict()
                ),
                "optimizer_state_dict": (
                    optimizer.state_dict()
                ),
                "val_loss": val_loss,
                "config": {
                    "scale_factor": SCALE_FACTOR,
                    "num_features": NUM_FEATURES,
                    "num_blocks": NUM_RES_BLOCKS
                }
            },
            BEST_MODEL_PATH
        )

        print(
            "  ✓ Best validation model saved"
        )

print("=" * 70)
print("TRAINING COMPLETE")
print("=" * 70)

print(
    "Best validation loss:",
    best_val_loss
)
```

```python
# ============================================================
# CELL 39 — TEST EDSR
# ============================================================

model.eval()

rmse_values = []

mae_values = []

all_gt = []

all_pred = []

slope_error_values = []

with torch.no_grad():

    for lr, gt, mean in test_loader:

        lr = lr.to(
            device,
            non_blocking=True
        )

        gt = gt.to(
            device,
            non_blocking=True
        )

        mean = mean.to(
            device,
            non_blocking=True
        )

        # ----------------------------------------------------
        # Super-resolution
        # ----------------------------------------------------

        sr = model(
            lr
        )

        # Safety check
        assert sr.shape[1] == 1

        # ----------------------------------------------------
        # IMPORTANT:
        # [B] → [B,1,1,1]
        # ----------------------------------------------------

        mean = mean.view(
            -1,
            1,
            1,
            1
        )

        # Restore elevation
        sr_dem = (
            sr +
            mean
        )

        gt_dem = (
            gt +
            mean
        )

        # ----------------------------------------------------
        # Height metrics
        # ----------------------------------------------------

        rmse, mae = (
            calculate_height_metrics(
                sr_dem,
                gt_dem
            )
        )

        rmse_values.append(
            rmse
        )

        mae_values.append(
            mae
        )

        # ----------------------------------------------------
        # Terrain analytics
        # ----------------------------------------------------

        slope_sr, roughness_sr = (
            analytics(
                sr_dem
            )
        )

        slope_gt, roughness_gt = (
            analytics(
                gt_dem
            )
        )

        # ----------------------------------------------------
        # Slope error
        # ----------------------------------------------------

        slope_error = torch.mean(
            torch.abs(
                slope_sr -
                slope_gt
            )
        )

        slope_error_values.append(
            slope_error.item()
        )

        # ----------------------------------------------------
        # Hazard
        # ----------------------------------------------------

        hazard_sr = hazard_classes(
            slope_sr
        )

        hazard_gt = hazard_classes(
            slope_gt
        )

        all_gt.extend(
            hazard_gt.cpu()
            .numpy()
            .ravel()
        )

        all_pred.extend(
            hazard_sr.cpu()
            .numpy()
            .ravel()
        )

print("✓ Test evaluation complete")
```

```python
import os
import time
import numpy as np
import streamlit as st
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

try:
    import rasterio
    from rasterio.io import MemoryFile
except Exception:
    rasterio = None

# ============================================================
# LUNAR-SAFE DASHBOARD — ROBUST VERSION
# ============================================================
st.set_page_config(
    page_title="LUNAR-SAFE | AI Hazard Analysis",
    page_icon="🌕",
    layout="wide",
    initial_sidebar_state="expanded",
)

st.markdown("""
<style>
.stApp { background:#080b14; color:#f4f5f7; }
.block-container { padding-top:1.2rem; max-width:1500px; }
.hero { padding:1.4rem 1.7rem; border:1px solid rgba(255,255,255,.12);
border-radius:20px; background:linear-gradient(135deg,#181d30,#0a0d18); margin-bottom:1rem; }
.hero h1 { margin:0; font-size:2.4rem; }
.hero p { color:#aeb7ca; margin:.45rem 0 0; }
.poc { border:1px solid rgba(96,165,250,.3); background:rgba(37,99,235,.08);
border-radius:16px; padding:1rem 1.1rem; margin:.5rem 0 1rem; }
</style>
""", unsafe_allow_html=True)


# ---------------- MODEL ----------------
class ResidualBlock(nn.Module):
    def __init__(self, channels=48, res_scale=0.1):
        super().__init__()
        self.res_scale = res_scale
        self.body = nn.Sequential(
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1),
        )

    def forward(self, x):
        return x + self.res_scale * self.body(x)


class FastEDSR(nn.Module):
    def __init__(self, scale_factor=5, num_res_blocks=4, num_features=48):
        super().__init__()
        self.scale_factor = scale_factor
        self.head = nn.Conv2d(1, num_features, 3, padding=1)
        self.body = nn.Sequential(
            *[ResidualBlock(num_features) for _ in range(num_res_blocks)]
        )
        self.body_tail = nn.Conv2d(num_features, num_features, 3, padding=1)
        self.residual_out = nn.Conv2d(num_features, 1, 3, padding=1)

    def forward(self, x):
        base = F.interpolate(
            x, scale_factor=5, mode="bilinear", align_corners=False
        )
        feat = self.head(x)
        res = self.body_tail(self.body(feat))
        res = self.residual_out(res)
        res = F.interpolate(
            res, scale_factor=5, mode="bilinear", align_corners=False
        )
        return base + res


def find_checkpoint(user_path):
    candidates = [
        user_path,
        "models/LUNAR_SAFE_V2_FINAL.pth",
        "./models/LUNAR_SAFE_V2_FINAL.pth",
        "/content/models/LUNAR_SAFE_V2_FINAL.pth",
        "/mnt/data/models/LUNAR_SAFE_V2_FINAL.pth",
        "LUNAR_SAFE_V2_FINAL.pth",
    ]
    for p in candidates:
        if p and os.path.isfile(p):
            return p
    return None


@st.cache_resource
def load_model(user_path):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    model = FastEDSR().to(device).eval()
    path = find_checkpoint(user_path)

    if path is None:
        return model, False, device, "Checkpoint not found"

    try:
        ckpt = torch.load(path, map_location=device)
        state = (
            ckpt["model_state_dict"]
            if isinstance(ckpt, dict) and "model_state_dict" in ckpt
            else ckpt
        )
        model.load_state_dict(state, strict=True)
        return model, True, device, f"Loaded: {path}"
    except Exception as e:
        return model, False, device, f"Checkpoint error: {e}"


# ---------------- DATA ----------------
def gaussian_kernel(k=5, sigma=1.0):
    r = k // 2
    ax = torch.arange(-r, r + 1, dtype=torch.float32)
    xx, yy = torch.meshgrid(ax, ax, indexing="ij")
    g = torch.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    return (g / g.sum()).view(1, 1, k, k)


def degrade_dem(hr, scale=5):
    x = torch.as_tensor(hr, dtype=torch.float32)[None, None]
    x = F.conv2d(x, gaussian_kernel(), padding=2)
    x = F.interpolate(x, scale_factor=1 / scale, mode="area")
    return x.squeeze().numpy()


def make_synthetic_dem(size=500, seed=42):
    rng = np.random.default_rng(seed)
    y, x = np.mgrid[0:size, 0:size]

    trend = .0025 * x + .0015 * y
    crater1 = -28 * np.exp(-((x-170)**2 + (y-190)**2) / (2*48**2))
    crater2 = -16 * np.exp(-((x-365)**2 + (y-325)**2) / (2*34**2))
    rim1 = 5 * np.exp(-((np.sqrt((x-170)**2+(y-190)**2)-52)**2) / (2*8**2))
    rim2 = 3 * np.exp(-((np.sqrt((x-365)**2+(y-325)**2)-38)**2) / (2*6**2))
    rough = rng.normal(0, .55, (size, size))

    return (
        1500 + 100*trend + crater1 + crater2 + rim1 + rim2 + rough
    ).astype(np.float32)


def center_crop(arr, h, w):
    H, W = arr.shape

    if H < h or W < w:
        t = torch.tensor(arr)[None, None].float()
        t = F.interpolate(
            t, size=(max(h, H), max(w, W)),
            mode="bilinear", align_corners=False
        )
        arr = t.squeeze().numpy()
        H, W = arr.shape

    y0, x0 = (H-h)//2, (W-w)//2
    return arr[y0:y0+h, x0:x0+w]


def read_uploaded(uploaded):
    if rasterio is None:
        raise RuntimeError("rasterio is not installed. Run: pip install rasterio")

    with MemoryFile(uploaded.getvalue()) as mem:
        with mem.open() as src:
            arr = src.read(1).astype(np.float32)

    arr = np.nan_to_num(arr, nan=float(np.nanmedian(arr)))
    return arr


# ---------------- TERRAIN ANALYSIS ----------------
def sobel_slope(dem):
    x = torch.tensor(dem, dtype=torch.float32)[None, None]

    kx = torch.tensor(
        [[-1,0,1],[-2,0,2],[-1,0,1]], dtype=torch.float32
    ) / 8

    ky = torch.tensor(
        [[-1,-2,-1],[0,0,0],[1,2,1]], dtype=torch.float32
    ) / 8

    gx = F.conv2d(x, kx[None,None], padding=1).squeeze().numpy()
    gy = F.conv2d(x, ky[None,None], padding=1).squeeze().numpy()

    return np.degrees(np.arctan(np.sqrt(gx*gx + gy*gy)))


def roughness(dem, window=5):
    x = torch.tensor(dem, dtype=torch.float32)[None, None]
    m = F.avg_pool2d(x, window, stride=1, padding=window//2)
    m2 = F.avg_pool2d(x*x, window, stride=1, padding=window//2)
    return torch.sqrt(torch.clamp(m2 - m*m, min=0)).squeeze().numpy()


# Dashboard demonstration thresholds from the existing prototype.
def hazard_map(slope, rough):
    out = np.zeros_like(slope, dtype=np.uint8)
    out[(slope >= 8) | (rough >= 1.5)] = 1
    out[(slope >= 15) | (rough >= 3.0)] = 2
    return out


# ---------------- PLOTS ----------------
def fig(arr, title, cmap="terrain", vmin=None, vmax=None):
    f, ax = plt.subplots(figsize=(6, 5))
    im = ax.imshow(arr, cmap=cmap, vmin=vmin, vmax=vmax)
    ax.set_title(title)
    ax.set_xticks([])
    ax.set_yticks([])
    f.colorbar(im, ax=ax, shrink=.8)
    f.tight_layout()
    return f


def hazard_fig(arr):
    f, ax = plt.subplots(figsize=(7, 5.5))
    cmap = ListedColormap(["#22c55e", "#facc15", "#ef4444"])
    im = ax.imshow(arr, cmap=cmap, vmin=0, vmax=2)
    ax.set_title("LUNAR-SAFE Terrain Hazard Classification")
    ax.set_xticks([])
    ax.set_yticks([])
    cb = f.colorbar(im, ax=ax, ticks=[0,1,2], shrink=.8)
    cb.ax.set_yticklabels(["Safe", "Moderate", "Hazardous"])
    f.tight_layout()
    return f


# ============================================================
# SIDEBAR
# ============================================================
st.sidebar.title("🌕 LUNAR-SAFE")

checkpoint_input = st.sidebar.text_input(
    "Model checkpoint",
    "models/LUNAR_SAFE_V2_FINAL.pth"
)

model, loaded, device, load_msg = load_model(checkpoint_input)

if loaded:
    st.sidebar.success("✓ Trained FastEDSR loaded")
else:
    st.sidebar.warning("⚠ Trained checkpoint not loaded")

st.sidebar.caption(load_msg)

mode = st.sidebar.radio(
    "Data mode",
    ["Synthetic POC Demo", "Upload LR GeoTIFF"]
)

uploaded = (
    st.sidebar.file_uploader("Upload GeoTIFF", type=["tif", "tiff"])
    if mode == "Upload LR GeoTIFF" else None
)

run = st.sidebar.button(
    "🚀 RUN LUNAR-SAFE",
    type="primary",
    use_container_width=True
)


# ============================================================
# HEADER
# ============================================================
st.markdown("""
<div class="hero">
<h1>🌕 LUNAR-SAFE</h1>
<p>AI-based DEM Super-Resolution → Terrain Analytics → Hazard Mapping</p>
</div>

<div class="poc">
<b>PHASE-I PROOF OF CONCEPT</b><br>
Low-resolution terrain → 5× reconstruction → slope & roughness → preliminary hazard classification.
</div>
""", unsafe_allow_html=True)


# ============================================================
# INPUT
# ============================================================
if mode == "Synthetic POC Demo":

    gt = center_crop(make_synthetic_dem(), 100, 100)
    lr = degrade_dem(gt)
    source = "Synthetic development terrain"

else:

    if uploaded is None:
        st.info("Upload a low-resolution GeoTIFF from the sidebar.")
        st.stop()

    try:
        raw = read_uploaded(uploaded)
        lr = center_crop(raw, 20, 20)
        gt = None
        source = uploaded.name
    except Exception as e:
        st.error(f"Could not read GeoTIFF: {e}")
        st.stop()


# ============================================================
# INFERENCE
# IMPORTANT: NEVER STOP THE DASHBOARD JUST BECAUSE CHECKPOINT
# IS MISSING. Show the dashboard and use bilinear as a fallback.
# ============================================================
x0 = torch.tensor(lr, dtype=torch.float32)
mean = float(x0.mean())
x = ((x0 - mean)[None, None]).to(device)

start = time.perf_counter()

if loaded:
    with torch.no_grad():
        if device.type == "cuda":
            with torch.autocast(device_type="cuda", dtype=torch.float16):
                sr_n = model(x)
            torch.cuda.synchronize()
        else:
            sr_n = model(x)

    sr = sr_n.squeeze().float().cpu().numpy() + mean
    inference_mode = "FastEDSR AI reconstruction"

else:
    # Safe visual fallback so the dashboard remains fully usable.
    # This is NOT presented as a trained-model result.
    sr = F.interpolate(
        torch.tensor(lr)[None,None].float(),
        size=(100,100),
        mode="bilinear",
        align_corners=False
    ).squeeze().numpy()
    inference_mode = "Bilinear demo fallback — trained checkpoint unavailable"

elapsed = (time.perf_counter() - start) * 1000

bilinear = F.interpolate(
    torch.tensor(lr)[None,None].float(),
    size=(100,100),
    mode="bilinear",
    align_corners=False
).squeeze().numpy()

slope = sobel_slope(sr)
rough = roughness(sr)
haz = hazard_map(slope, rough)


# ============================================================
# STATUS
# ============================================================
if loaded:
    st.success("MODEL OPERATIONAL — FastEDSR V2 checkpoint is loaded.")
else:
    st.warning(
        "Dashboard is running in DEMO mode. The trained FastEDSR checkpoint "
        "was not found, so the reconstruction shown below uses bilinear "
        "interpolation. Add LUNAR_SAFE_V2_FINAL.pth to enable AI inference."
    )

st.caption(f"Input source: {source}  |  Processing mode: {inference_mode}")


# ============================================================
# METRICS
# ============================================================
st.markdown("### Mission Processing Status")

c1, c2, c3, c4, c5 = st.columns(5)

c1.metric("Model", "FastEDSR ×5" if loaded else "Demo")
c2.metric("Input", f"{lr.shape[0]} × {lr.shape[1]}")
c3.metric("Output", f"{sr.shape[0]} × {sr.shape[1]}")
c4.metric("Scale", "5×")
c5.metric("Inference", f"{elapsed:.1f} ms")


# ============================================================
# TABS
# ============================================================
t1, t2, t3, t4 = st.tabs([
    "🛰️ Reconstruction",
    "⛰️ Terrain Analytics",
    "⚠️ Hazard Map",
    "📊 Performance"
])


with t1:

    a, b, c = st.columns(3)

    with a:
        st.pyplot(
            fig(lr, "Input Low-Resolution DEM"),
            use_container_width=True
        )

    with b:
        st.pyplot(
            fig(bilinear, "Bilinear Baseline"),
            use_container_width=True
        )

    with c:
        st.pyplot(
            fig(sr, "LUNAR-SAFE Reconstruction"),
            use_container_width=True
        )

    if gt is not None:

        st.markdown("#### High-Resolution Reference")

        a, b = st.columns(2)

        with a:
            st.pyplot(
                fig(gt, "Ground Truth"),
                use_container_width=True
            )

        with b:
            st.pyplot(
                fig(np.abs(sr-gt), "Absolute Reconstruction Error", "magma"),
                use_container_width=True
            )

        rmse = float(np.sqrt(np.mean((sr-gt)**2)))
        mae = float(np.mean(np.abs(sr-gt)))

        q1, q2, q3 = st.columns(3)

        q1.metric("Current Reconstruction RMSE", f"{rmse:.4f} m")
        q2.metric("Current Reconstruction MAE", f"{mae:.4f} m")
        q3.metric("Bilinear RMSE", f"{np.sqrt(np.mean((bilinear-gt)**2)):.4f} m")


with t2:

    a, b = st.columns(2)

    with a:
        st.pyplot(
            fig(slope, "Derived Slope", "viridis"),
            use_container_width=True
        )

    with b:
        st.pyplot(
            fig(rough, "Terrain Roughness", "plasma"),
            use_container_width=True
        )

    a, b, c = st.columns(3)

    a.metric("Mean Slope", f"{np.mean(slope):.2f}°")
    b.metric("Max Slope", f"{np.max(slope):.2f}°")
    c.metric("Mean Roughness", f"{np.mean(rough):.3f} m")

    st.info(
        "Slope and roughness are calculated from the reconstructed DEM."
    )


with t3:

    st.pyplot(
        hazard_fig(haz),
        use_container_width=True
    )

    safe = 100*np.mean(haz == 0)
    moderate = 100*np.mean(haz == 1)
    dangerous = 100*np.mean(haz == 2)

    a, b, c = st.columns(3)

    a.metric("SAFE TERRAIN", f"{safe:.1f}%")
    b.metric("MODERATE TERRAIN", f"{moderate:.1f}%")
    c.metric("HAZARDOUS TERRAIN", f"{dangerous:.1f}%")

    st.markdown("#### Dashboard Prototype Decision Rules")

    st.code(
"""SAFE       : slope < 8° AND roughness < 1.5 m
MODERATE   : slope < 15° AND roughness < 3.0 m
HAZARDOUS  : otherwise"""
    )

    st.warning(
        "These dashboard thresholds are demonstration rules, not certified "
        "lunar landing safety limits."
    )


with t4:

    st.markdown("### Performance")

    if gt is not None:

        gt_slope = sobel_slope(gt)
        gt_h = hazard_map(gt_slope, roughness(gt))

        acc = 100*np.mean(haz == gt_h)
        slope_mae = float(np.mean(np.abs(slope-gt_slope)))

        a, b, c = st.columns(3)

        a.metric("Demo Hazard Accuracy", f"{acc:.2f}%")
        b.metric("Slope MAE", f"{slope_mae:.3f}°")
        c.metric("Inference", f"{elapsed:.1f} ms")

        st.dataframe({
            "Metric": [
                "RMSE (m)",
                "MAE (m)",
                "Slope MAE (°)",
                "Hazard Accuracy (%)"
            ],
            "Current Dashboard": [
                float(np.sqrt(np.mean((sr-gt)**2))),
                float(np.mean(np.abs(sr-gt))),
                slope_mae,
                acc
            ]
        }, hide_index=True, use_container_width=True)

    else:

        st.info(
            "No high-resolution reference was supplied, so supervised "
            "accuracy metrics are not computed."
        )

        st.metric("Inference Time", f"{elapsed:.1f} ms")


# ============================================================
# FOOTER
# ============================================================
st.markdown("---")

st.markdown("""
<div style="text-align:center;color:#8993a8;padding:1rem">
<b style="color:#dbe4ff">LUNAR-SAFE</b> —
AI Lunar Terrain Super-Resolution & Hazard Analysis<br>
<span style="font-size:.8rem">
Phase-I proof-of-concept research dashboard • Development data •
Not for real mission navigation
</span>
</div>
""", unsafe_allow_html=True)
```

```python
# ============================================================
# 🌕 LUNAR-SAFE — INLINE COLAB DEMO
# No Streamlit | No localhost | No separate website
# Everything appears directly below this cell
# ============================================================

import os
import time
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from IPython.display import display, HTML

# ------------------------------------------------------------
# 1. CONFIGURATION
# ------------------------------------------------------------

MODEL_PATH = "models/LUNAR_SAFE_V2_FINAL.pth"

SCALE = 5
HR_SIZE = 100
LR_SIZE = 20

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("🌕 LUNAR-SAFE — AI Lunar Terrain Analysis")
print("=" * 60)
print(f"Device : {device}")
print(f"Scale  : {SCALE}×")
print(f"Input  : {LR_SIZE} × {LR_SIZE}")
print(f"Output : {HR_SIZE} × {HR_SIZE}")
print("=" * 60)


# ------------------------------------------------------------
# 2. FASTEDSR MODEL
# ------------------------------------------------------------

class ResidualBlock(nn.Module):

    def __init__(self, channels=48, res_scale=0.1):
        super().__init__()

        self.res_scale = res_scale

        self.body = nn.Sequential(
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(channels, channels, 3, padding=1)
        )

    def forward(self, x):

        return x + self.res_scale * self.body(x)


class FastEDSR(nn.Module):

    def __init__(
        self,
        scale_factor=5,
        num_res_blocks=4,
        num_features=48
    ):

        super().__init__()

        self.scale_factor = scale_factor

        # Input feature extraction
        self.head = nn.Conv2d(
            1,
            num_features,
            3,
            padding=1
        )

        # Residual blocks
        self.body = nn.Sequential(
            *[
                ResidualBlock(num_features)
                for _ in range(num_res_blocks)
            ]
        )

        # Body convolution
        self.body_tail = nn.Conv2d(
            num_features,
            num_features,
            3,
            padding=1
        )

        # Generate terrain residual
        self.residual_out = nn.Conv2d(
            num_features,
            1,
            3,
            padding=1
        )


    def forward(self, x):

        # Bilinear baseline
        base = F.interpolate(
            x,
            scale_factor=5,
            mode="bilinear",
            align_corners=False
        )

        # Feature extraction
        features = self.head(x)

        # Residual learning
        residual = self.body(features)

        residual = self.body_tail(residual)

        residual = self.residual_out(residual)

        # Upsample learned terrain residual
        residual = F.interpolate(
            residual,
            scale_factor=5,
            mode="bilinear",
            align_corners=False
        )

        # Final reconstruction
        output = base + residual

        return output


# ------------------------------------------------------------
# 3. LOAD TRAINED MODEL
# ------------------------------------------------------------

model = FastEDSR().to(device)
model.eval()

model_loaded = False

possible_paths = [
    MODEL_PATH,
    "/content/models/LUNAR_SAFE_V2_FINAL.pth",
    "/content/LUNAR_SAFE_V2_FINAL.pth",
    "LUNAR_SAFE_V2_FINAL.pth"
]

for path in possible_paths:

    if os.path.exists(path):

        try:

            checkpoint = torch.load(
                path,
                map_location=device
            )

            if isinstance(checkpoint, dict) and \
               "model_state_dict" in checkpoint:

                state_dict = checkpoint["model_state_dict"]

            else:

                state_dict = checkpoint

            model.load_state_dict(
                state_dict,
                strict=True
            )

            model_loaded = True

            print(f"✅ FastEDSR model loaded:")
            print(f"   {path}")

            break

        except Exception as e:

            print("⚠ Model found but could not be loaded:")
            print(e)


if not model_loaded:

    print()
    print("⚠ WARNING")
    print("Trained LUNAR_SAFE_V2_FINAL.pth was not found.")
    print("The demonstration will use bilinear interpolation.")
    print("Place the .pth file in:")
    print("models/LUNAR_SAFE_V2_FINAL.pth")


# ------------------------------------------------------------
# 4. GENERATE SYNTHETIC LUNAR-LIKE DEM
# ------------------------------------------------------------

np.random.seed(42)

SIZE = 500

y, x = np.mgrid[0:SIZE, 0:SIZE]


# Large-scale terrain trend
terrain_trend = (
    0.0025 * x +
    0.0015 * y
)


# Crater 1
crater1 = -28 * np.exp(
    -(
        (x - 170)**2 +
        (y - 190)**2
    ) /
    (2 * 48**2)
)


# Crater 2
crater2 = -16 * np.exp(
    -(
        (x - 365)**2 +
        (y - 325)**2
    ) /
    (2 * 34**2)
)


# Crater rims
rim1 = 5 * np.exp(
    -(
        (
            np.sqrt(
                (x-170)**2 +
                (y-190)**2
            ) - 52
        )**2
    ) /
    (2 * 8**2)
)


rim2 = 3 * np.exp(
    -(
        (
            np.sqrt(
                (x-365)**2 +
                (y-325)**2
            ) - 38
        )**2
    ) /
    (2 * 6**2)
)


# Small-scale terrain roughness
roughness_noise = np.random.normal(
    0,
    0.55,
    (SIZE, SIZE)
)


# Final synthetic DEM
synthetic_dem = (
    1500
    + 100 * terrain_trend
    + crater1
    + crater2
    + rim1
    + rim2
    + roughness_noise
).astype(np.float32)


# Center crop to 100×100
start_y = (SIZE - HR_SIZE) // 2
start_x = (SIZE - HR_SIZE) // 2

hr_dem = synthetic_dem[
    start_y:start_y+HR_SIZE,
    start_x:start_x+HR_SIZE
]


# ------------------------------------------------------------
# 5. CREATE LOW-RESOLUTION INPUT
# ------------------------------------------------------------

hr_tensor = torch.tensor(
    hr_dem,
    dtype=torch.float32
)[None, None]


# Gaussian blur
kernel_size = 5
sigma = 1.0

radius = kernel_size // 2

ax = torch.arange(
    -radius,
    radius + 1,
    dtype=torch.float32
)

xx, yy = torch.meshgrid(
    ax,
    ax,
    indexing="ij"
)

kernel = torch.exp(
    -(xx**2 + yy**2) /
    (2 * sigma**2)
)

kernel = kernel / kernel.sum()

kernel = kernel.view(
    1, 1,
    kernel_size,
    kernel_size
)


blurred = F.conv2d(
    hr_tensor,
    kernel,
    padding=radius
)


# 5× downsampling
lr_tensor = F.interpolate(
    blurred,
    size=(LR_SIZE, LR_SIZE),
    mode="area"
)

lr_dem = lr_tensor.squeeze().numpy()


# ------------------------------------------------------------
# 6. NORMALIZE INPUT
# ------------------------------------------------------------

input_tensor = torch.tensor(
    lr_dem,
    dtype=torch.float32
)

mean_elevation = input_tensor.mean()

normalized_input = (
    input_tensor - mean_elevation
)[None, None].to(device)


# ------------------------------------------------------------
# 7. FASTEDSR INFERENCE
# ------------------------------------------------------------

start_time = time.perf_counter()


if model_loaded:

    with torch.no_grad():

        sr_output = model(
            normalized_input
        )

    sr_dem = (
        sr_output
        .squeeze()
        .cpu()
        .numpy()
        + mean_elevation.item()
    )

else:

    # Bilinear fallback
    sr_dem = F.interpolate(
        torch.tensor(lr_dem)[None, None],
        size=(HR_SIZE, HR_SIZE),
        mode="bilinear",
        align_corners=False
    ).squeeze().numpy()


inference_time = (
    time.perf_counter() - start_time
) * 1000


# ------------------------------------------------------------
# 8. BILINEAR BASELINE
# ------------------------------------------------------------

bilinear_dem = F.interpolate(
    torch.tensor(lr_dem)[None, None],
    size=(HR_SIZE, HR_SIZE),
    mode="bilinear",
    align_corners=False
).squeeze().numpy()


# ------------------------------------------------------------
# 9. TERRAIN SLOPE
# ------------------------------------------------------------

def calculate_slope(dem):

    dem_tensor = torch.tensor(
        dem,
        dtype=torch.float32
    )[None, None]

    sobel_x = torch.tensor(
        [
            [-1, 0, 1],
            [-2, 0, 2],
            [-1, 0, 1]
        ],
        dtype=torch.float32
    ) / 8

    sobel_y = torch.tensor(
        [
            [-1, -2, -1],
            [ 0,  0,  0],
            [ 1,  2,  1]
        ],
        dtype=torch.float32
    ) / 8

    sobel_x = sobel_x[None, None]

    sobel_y = sobel_y[None, None]

    gx = F.conv2d(
        dem_tensor,
        sobel_x,
        padding=1
    )

    gy = F.conv2d(
        dem_tensor,
        sobel_y,
        padding=1
    )

    gradient = torch.sqrt(
        gx**2 + gy**2
    )

    slope = torch.atan(
        gradient
    ) * 180 / np.pi

    return slope.squeeze().numpy()


slope = calculate_slope(
    sr_dem
)


# ------------------------------------------------------------
# 10. TERRAIN ROUGHNESS
# ------------------------------------------------------------

def calculate_roughness(
    dem,
    window=5
):

    tensor = torch.tensor(
        dem,
        dtype=torch.float32
    )[None, None]

    mean = F.avg_pool2d(
        tensor,
        window,
        stride=1,
        padding=window//2
    )

    mean_squared = F.avg_pool2d(
        tensor**2,
        window,
        stride=1,
        padding=window//2
    )

    variance = (
        mean_squared -
        mean**2
    )

    rough = torch.sqrt(
        torch.clamp(
            variance,
            min=0
        )
    )

    return rough.squeeze().numpy()


terrain_roughness = calculate_roughness(
    sr_dem
)


# ------------------------------------------------------------
# 11. HAZARD CLASSIFICATION
# ------------------------------------------------------------

# Dashboard demonstration thresholds
# Safe       : slope < 8° and roughness < 1.5
# Moderate   : slope 8–15° or roughness 1.5–3
# Hazardous  : slope >=15° or roughness >=3

hazard_map = np.zeros_like(
    slope,
    dtype=np.uint8
)

hazard_map[
    (slope >= 8) |
    (terrain_roughness >= 1.5)
] = 1

hazard_map[
    (slope >= 15) |
    (terrain_roughness >= 3)
] = 2


# ------------------------------------------------------------
# 12. RECONSTRUCTION METRICS
# ------------------------------------------------------------

rmse = np.sqrt(
    np.mean(
        (sr_dem - hr_dem)**2
    )
)

mae = np.mean(
    np.abs(
        sr_dem - hr_dem
    )
)

bilinear_rmse = np.sqrt(
    np.mean(
        (bilinear_dem - hr_dem)**2
    )
)

bilinear_mae = np.mean(
    np.abs(
        bilinear_dem - hr_dem
    )
)


# ------------------------------------------------------------
# 13. DISPLAY HEADER
# ------------------------------------------------------------

display(
    HTML(
        """
        <div style="
        background:linear-gradient(135deg,#111827,#1e293b);
        padding:25px;
        border-radius:15px;
        color:white;
        margin-bottom:20px;">

        <h1 style="margin:0;">
        🌕 LUNAR-SAFE
        </h1>

        <p style="font-size:17px;">
        AI-Based Lunar DEM Super-Resolution
        & Preliminary Terrain Hazard Analysis
        </p>

        <p>
        <b>Pipeline:</b>
        Low-Resolution DEM → FastEDSR → High-Resolution DEM
        → Slope → Roughness → Hazard Map
        </p>

        </div>
        """
    )
)


# ------------------------------------------------------------
# 14. STATUS
# ------------------------------------------------------------

if model_loaded:

    display(
        HTML(
            """
            <div style="
            padding:15px;
            border-radius:10px;
            background:#064e3b;
            color:white;">
            ✅ <b>FASTEDSR MODEL LOADED</b><br>
            Trained LUNAR-SAFE V2 checkpoint is being used.
            </div>
            """
        )
    )

else:

    display(
        HTML(
            """
            <div style="
            padding:15px;
            border-radius:10px;
            background:#78350f;
            color:white;">
            ⚠ <b>DEMO MODE</b><br>
            Trained checkpoint was not found.
            Bilinear interpolation is being shown as fallback.
            </div>
            """
        )
    )


# ------------------------------------------------------------
# 15. MAIN DEM VISUALIZATION
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1,
    4,
    figsize=(20, 5)
)


im0 = axes[0].imshow(
    lr_dem,
    cmap="terrain"
)

axes[0].set_title(
    "Input DEM\n20 × 20"
)

axes[0].axis("off")

plt.colorbar(
    im0,
    ax=axes[0],
    fraction=.046
)


im1 = axes[1].imshow(
    bilinear_dem,
    cmap="terrain"
)

axes[1].set_title(
    "Bilinear Baseline\n100 × 100"
)

axes[1].axis("off")

plt.colorbar(
    im1,
    ax=axes[1],
    fraction=.046
)


im2 = axes[2].imshow(
    sr_dem,
    cmap="terrain"
)

axes[2].set_title(
    "LUNAR-SAFE FastEDSR\n100 × 100"
)

axes[2].axis("off")

plt.colorbar(
    im2,
    ax=axes[2],
    fraction=.046
)


im3 = axes[3].imshow(
    np.abs(sr_dem - hr_dem),
    cmap="magma"
)

axes[3].set_title(
    "Reconstruction Error"
)

axes[3].axis("off")

plt.colorbar(
    im3,
    ax=axes[3],
    fraction=.046
)


plt.suptitle(
    "DEM Super-Resolution Results",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 16. TERRAIN ANALYSIS
# ------------------------------------------------------------

fig, axes = plt.subplots(
    1,
    3,
    figsize=(17, 5)
)


im1 = axes[0].imshow(
    sr_dem,
    cmap="terrain"
)

axes[0].set_title(
    "Reconstructed Elevation"
)

axes[0].axis("off")

plt.colorbar(
    im1,
    ax=axes[0],
    fraction=.046
)


im2 = axes[1].imshow(
    slope,
    cmap="viridis"
)

axes[1].set_title(
    "Slope Map (degrees)"
)

axes[1].axis("off")

plt.colorbar(
    im2,
    ax=axes[1],
    fraction=.046
)


im3 = axes[2].imshow(
    terrain_roughness,
    cmap="plasma"
)

axes[2].set_title(
    "Surface Roughness"
)

axes[2].axis("off")

plt.colorbar(
    im3,
    ax=axes[2],
    fraction=.046
)


plt.suptitle(
    "Terrain Analytics",
    fontsize=18,
    fontweight="bold"
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 17. HAZARD MAP
# ------------------------------------------------------------

from matplotlib.colors import ListedColormap

hazard_colors = [
    "#22c55e",
    "#facc15",
    "#ef4444"
]

hazard_cmap = ListedColormap(
    hazard_colors
)

plt.figure(
    figsize=(8, 6)
)

plt.imshow(
    hazard_map,
    cmap=hazard_cmap,
    vmin=0,
    vmax=2
)

plt.title(
    "🌕 LUNAR-SAFE Preliminary Terrain Hazard Map",
    fontsize=16,
    fontweight="bold"
)

plt.axis("off")

cbar = plt.colorbar(
    ticks=[0,1,2],
    fraction=.046
)

cbar.ax.set_yticklabels(
    [
        "SAFE",
        "MODERATE",
        "HAZARDOUS"
    ]
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 18. FINAL METRICS DASHBOARD
# ------------------------------------------------------------

safe_percentage = (
    np.mean(hazard_map == 0) * 100
)

moderate_percentage = (
    np.mean(hazard_map == 1) * 100
)

hazardous_percentage = (
    np.mean(hazard_map == 2) * 100
)


display(
    HTML(
        f"""
        <div style="
        background:#111827;
        padding:20px;
        border-radius:15px;
        color:white;
        margin-top:15px;">

        <h2>📊 LUNAR-SAFE Results</h2>

        <table style="
        width:100%;
        border-collapse:collapse;
        font-size:16px;">

        <tr>
        <td><b>Model</b></td>
        <td>{"FastEDSR V2" if model_loaded else "Bilinear Demo Fallback"}</td>
        </tr>

        <tr>
        <td><b>Input Resolution</b></td>
        <td>20 × 20</td>
        </tr>

        <tr>
        <td><b>Output Resolution</b></td>
        <td>100 × 100</td>
        </tr>

        <tr>
        <td><b>Super-Resolution Scale</b></td>
        <td>5×</td>
        </tr>

        <tr>
        <td><b>RMSE</b></td>
        <td>{rmse:.4f} m</td>
        </tr>

        <tr>
        <td><b>MAE</b></td>
        <td>{mae:.4f} m</td>
        </tr>

        <tr>
        <td><b>Bilinear RMSE</b></td>
        <td>{bilinear_rmse:.4f} m</td>
        </tr>

        <tr>
        <td><b>Bilinear MAE</b></td>
        <td>{bilinear_mae:.4f} m</td>
        </tr>

        <tr>
        <td><b>Mean Slope</b></td>
        <td>{np.mean(slope):.2f}°</td>
        </tr>

        <tr>
        <td><b>Mean Roughness</b></td>
        <td>{np.mean(terrain_roughness):.4f}</td>
        </tr>

        <tr>
        <td><b>Safe Terrain</b></td>
        <td>{safe_percentage:.2f}%</td>
        </tr>

        <tr>
        <td><b>Moderate Terrain</b></td>
        <td>{moderate_percentage:.2f}%</td>
        </tr>

        <tr>
        <td><b>Hazardous Terrain</b></td>
        <td>{hazardous_percentage:.2f}%</td>
        </tr>

        <tr>
        <td><b>Inference Time</b></td>
        <td>{inference_time:.3f} ms</td>
        </tr>

        </table>

        </div>
        """
    )
)


# ------------------------------------------------------------
# 19. PROJECT PIPELINE
# ------------------------------------------------------------

display(
    HTML(
        """
        <div style="
        margin-top:20px;
        padding:20px;
        background:#172033;
        border-radius:15px;
        color:white;
        text-align:center;">

        <h3>🔬 Complete LUNAR-SAFE Pipeline</h3>

        <p style="font-size:18px;">
        🛰️ Low-Resolution DEM
        →
        🧠 FastEDSR 5×
        →
        🗺️ High-Resolution DEM
        →
        ⛰️ Slope + Roughness
        →
        ⚠️ Hazard Classification
        →
        🌕 Preliminary Hazard Map
        </p>

        <p style="color:#9ca3af;">
        Phase-I research prototype • Synthetic development data •
        Not for real lunar mission navigation
        </p>

        </div>
        """
    )
)
```
